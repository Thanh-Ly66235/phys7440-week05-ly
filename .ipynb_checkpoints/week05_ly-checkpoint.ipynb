This notebook builds a small ReAct-style agent (Yao et al., arXiv:2210.03629). A language model reads a compound physics question, decides which tool to call, reads the result and repeats until it can answer. The tools live in a separate process, the MCP server `mcp_server_week05_ly.py`. The model client and the server communicate only through the Model Context Protocol, i.e. JSON-RPC 2.0 messages over the server's stdin/stdout:

```
Claude (Anthropic API) <-- tool_use / tool_result --> agent loop in this notebook (MCP client)
                                                                 |  JSON-RPC 2.0 over stdio
                                                                 v
                                            mcp_server_week05_ly.py  (subprocess)
                                              |-- query_simbad --> SIMBAD TAP service (astroquery.simbad)
                                              '-- search_arxiv --> arXiv API (arxiv package)
```

| 0 | Setup: imports, library versions, configuration, API key from `.env` |
| 1 | Start the MCP server as a subprocess, inspect the tool schemas, call the tools directly |
| 2 | The agentic loop, with every tool call logged |
| 3 | Betelgeuse: the compound query from the assignment |
| 4 | Second object: Proxima Centauri |
| 5 | Physical plausibility: SIMBAD web cross-check, sanity checks on physical quantities, summary vs. raw abstracts |

## 1. Start the MCP server and inspect its tools

**Purpose:** confirm that the server starts as a subprocess, completes the MCP handshake and advertises complete JSON Schemas, before any LLM is involved.

`mcp_session()` launches `python mcp_server_week05_ly.py` with the interpreter that runs this kernel (`sys.executable`) and the notebook folder as working directory, so no path has to be edited. Client and server then exchange JSON-RPC requests: `initialize` (protocol version and capabilities), `tools/list` (tool names, descriptions and schemas) and one `tools/call` per tool invocation. Two practical details:

* The server's **stderr** goes to `logs/mcp_server_stderr.log`. Jupyter's own `sys.stderr` is not a real file, and the subprocess needs one.
* The server inherits this kernel's environment **minus every `*_API_KEY` variable**. It does not need the Anthropic key, and keeping settings such as `HTTPS_PROXY` lets it work on networks that require a proxy.

Each agent run opens its own session, so one server process is started and stopped per question. This avoids keeping a connection open across notebook cells, which the transport does not support cleanly.

````python
def server_parameters(extra_env: dict[str, str] | None = None) -> StdioServerParameters:
    """Command that launches the MCP server; API keys are withheld from the subprocess."""
    env = {key: value for key, value in os.environ.items() if not key.endswith("_API_KEY")}
    env.update(extra_env or {})
    return StdioServerParameters(command=sys.executable, args=[str(SERVER_SCRIPT)], env=env, cwd=str(PROJECT_DIR))


@asynccontextmanager
async def mcp_session(extra_env: dict[str, str] | None = None):
    """Start the MCP server subprocess, perform the initialize handshake and yield a live ClientSession."""
    with SERVER_LOG.open("a", encoding="utf-8") as errlog:
        async with stdio_client(server_parameters(extra_env), errlog=errlog) as (read_stream, write_stream):
            async with ClientSession(read_stream, write_stream) as session:
                await session.initialize()
                yield session


def result_payload(result):
    """Return an MCP tool result as Python data: structured content if present, otherwise the text."""
    if result.structured_content is not None:
        data = result.structured_content
        return data["result"] if set(data) == {"result"} else data   # list outputs are wrapped as {"result": [...]}
    return "\n".join(block.text for block in result.content if getattr(block, "type", "") == "text")


def to_json(data, max_chars: int | None = None) -> str:
    """Pretty-print data as JSON (strings unchanged), optionally truncated for display."""
    text = data if isinstance(data, str) else json.dumps(data, indent=2, ensure_ascii=False)
    if max_chars is not None and len(text) > max_chars:
        return text[:max_chars] + f" ... [{len(text) - max_chars} more characters]"
    return text
````

````python
async with mcp_session() as session:
    handshake = session.initialize_result      # result of the initialize handshake done in mcp_session()
    mcp_tools = (await session.list_tools()).tools

print(f"Server: {handshake.server_info.name} v{handshake.server_info.version}")
print(f"MCP protocol version negotiated: {handshake.protocol_version}")
print(f"Tools advertised: {[tool.name for tool in mcp_tools]}\n")
for tool in mcp_tools:
    print(f"========== {tool.name} ==========")
    print(tool.description.strip(), "\n")
    print("inputSchema =", json.dumps(tool.input_schema, indent=2))
    output_properties = tool.output_schema.get("properties", {})
    if "result" in output_properties:                                  # list output: describe the item type
        item_ref = output_properties["result"]["items"]["$ref"].split("/")[-1]
        output_properties = tool.output_schema["$defs"][item_ref]["properties"]
    print(f"outputSchema fields: {list(output_properties)}\n")
````

**Schema completeness check.** The LLM sees only a tool's name, its description and its input schema, so every argument needs a type, a human-readable description and explicit bounds. The cell below checks this programmatically. The schemas are generated by the `mcp` SDK from the type hints and `pydantic.Field(...)` annotations in the server module.

````python
def schema_problems(tool) -> list[str]:
    """List the ways a tool definition falls short of being self-explanatory to an LLM."""
    problems = []
    schema = tool.input_schema
    if len(tool.description or "") < 80:
        problems.append("description shorter than 80 characters")
    if schema.get("type") != "object":
        problems.append("input schema is not an object")
    properties = schema.get("properties", {})
    if set(schema.get("required", [])) != set(properties):
        problems.append("not every argument is required")
    for name, prop in properties.items():
        problems += [f"'{name}' has no {key}" for key in ("type", "description", "examples") if key not in prop]
        if prop.get("type") == "integer" and not {"minimum", "maximum"} <= prop.keys():
            problems.append(f"'{name}' has no minimum/maximum")
        if prop.get("type") == "string" and not {"minLength", "maxLength"} <= prop.keys():
            problems.append(f"'{name}' has no minLength/maxLength")
    if not tool.output_schema:
        problems.append("no output schema")
    return problems


schema_report = pd.DataFrame([{
    "tool": tool.name,
    "arguments (type, bounds)": "; ".join(
        f"{name}: {prop['type']} [{prop.get('minimum', prop.get('minLength'))}, {prop.get('maximum', prop.get('maxLength'))}]"
        for name, prop in tool.input_schema["properties"].items()),
    "required": ", ".join(tool.input_schema.get("required", [])),
    "output schema": "yes" if tool.output_schema else "no",
    "problems": "; ".join(schema_problems(tool)) or "none",
} for tool in mcp_tools])
schemas_complete = all(schema_report["problems"] == "none")
print(f"All tool schemas complete: {schemas_complete}")
schema_report
````

### Direct tool calls (no LLM)

**Purpose:** check that both tools work and that the server enforces its own schema, independently of any model. The last two calls send deliberately invalid arguments. The server should reject them with `isError: true` instead of crashing or querying the remote service.

````python
DIRECT_CALLS = [
    ("query_simbad", {"object_name": FIRST_OBJECT}),
    ("search_arxiv", {"keyword": FIRST_OBJECT, "max_results": 2}),
    ("search_arxiv", {"keyword": FIRST_OBJECT, "max_results": 500}),   # violates maximum = 25
    ("query_simbad", {"object_name": ""}),                              # violates minLength = 1
]

direct_results = []
async with mcp_session() as session:
    for tool_name, arguments in DIRECT_CALLS:
        started = time.perf_counter()
        result = await session.call_tool(tool_name, arguments)
        direct_results.append({"tool": tool_name, "arguments": arguments, "is_error": bool(result.is_error),
                               "payload": result_payload(result)})
        print(f"tools/call {tool_name}({json.dumps(arguments)}) -> isError = {result.is_error} "
              f"({time.perf_counter() - started:.1f} s)")
        print(textwrap.indent(to_json(result_payload(result), max_chars=1500), "    "), "\n")

invalid_calls_rejected = all(call["is_error"] for call in direct_results[2:])
valid_calls_succeeded = not any(call["is_error"] for call in direct_results[:2])
print(f"Valid calls succeeded: {valid_calls_succeeded} | invalid calls rejected with isError: {invalid_calls_rejected}")
````

The valid calls return JSON whose fields match the output schema (missing catalogue values would be `null`, not omitted). The two invalid calls never reach SIMBAD or arXiv: the SDK validates the arguments against the input schema with `pydantic` first, and the client receives an `isError` result whose text names the violated constraint (`less_than_equal` for `max_results = 500`, `string_too_short` for the empty name). This is also how the agent receives errors: the message is passed back to the model, which can correct its arguments (Section 6).

## 2. The agentic loop

**Purpose:** connect Claude to the MCP tools and let the model decide which tools to call, in which order and with which arguments.

The loop follows the ReAct pattern:

1. Send the question, a short system prompt and the tool list to Claude. MCP's `inputSchema` is passed unchanged as Anthropic's `input_schema`.
2. If the reply stops with `stop_reason == "tool_use"`, execute every requested call through the MCP session (`tools/call`). Log the tool name, input arguments and returned result, then send the results back as `tool_result` blocks, with `is_error` set when the server reported an error.
3. Repeat until Claude answers without requesting a tool, or until `MAX_AGENT_TURNS` model calls have been made.

Every run is saved as JSON in `agent_runs/`: the question, the model, the model's text at each turn, every tool call with its full result, the final answer and the token usage. A later execution of the notebook reloads that file instead of calling the API, so the notebook runs top to bottom without an API key and its outputs stay identical to the analysed run. Set `FORCE_RERUN = True` to query the live services again; arXiv will then return newer papers.

````python
SYSTEM_PROMPT = (
    "You are an astrophysics research assistant. You can call tools that query the SIMBAD astronomical "
    "database and the arXiv preprint server. Use the tools for every catalogue value and every paper you "
    "mention instead of answering from memory, and quote numbers with the units and uncertainties the tools "
    "return. When you summarise papers, rely on the abstracts returned by the tool, cite each paper by its "
    "arXiv ID, and say explicitly when the abstracts do not address part of the question."
)


def to_anthropic_tools(tools) -> list[dict]:
    """Translate MCP tool definitions into the Anthropic Messages API tool format."""
    return [{"name": tool.name, "description": tool.description, "input_schema": tool.input_schema}
            for tool in tools]


async def execute_tool_call(session: ClientSession, name: str, arguments: dict, turn: int) -> dict:
    """Run one tools/call request and return its log entry; transport failures are recorded, not raised."""
    started = time.perf_counter()
    try:
        result = await session.call_tool(name, arguments)
        payload, is_error = result_payload(result), bool(result.is_error)
    except Exception as exc:          # e.g. the server process died during the call
        payload, is_error = f"MCP call failed: {type(exc).__name__}: {exc}", True
    return {"turn": turn, "tool": name, "arguments": arguments, "is_error": is_error,
            "seconds": round(time.perf_counter() - started, 2), "result": payload}


async def run_agent(question: str, *, max_turns: int = MAX_AGENT_TURNS,
                    server_env: dict[str, str] | None = None) -> dict:
    """Answer a question with Claude + the MCP tools and return a JSON-serialisable record of the run."""
    record = {"question": question, "model": LLM_MODEL, "max_turns": max_turns,
              "system_prompt": SYSTEM_PROMPT, "server_env": server_env or {},
              "started_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"),
              "assistant_text": [], "tool_calls": [], "final_answer": None, "status": None,
              "llm_calls": 0, "input_tokens": 0, "output_tokens": 0}
    messages = [{"role": "user", "content": question}]
    async with anthropic.AsyncAnthropic(max_retries=6) as llm, mcp_session(server_env) as session:
        tools = to_anthropic_tools((await session.list_tools()).tools)
        for turn in range(1, max_turns + 1):
            response = await llm.messages.create(model=LLM_MODEL, max_tokens=LLM_MAX_TOKENS, system=SYSTEM_PROMPT,
                                                 tools=tools, messages=messages)
            record["llm_calls"] += 1
            record["input_tokens"] += response.usage.input_tokens
            record["output_tokens"] += response.usage.output_tokens
            messages.append({"role": "assistant", "content": response.content})

            text = "\n".join(block.text for block in response.content if block.type == "text").strip()
            if text:
                record["assistant_text"].append({"turn": turn, "text": text})
            tool_uses = [block for block in response.content if block.type == "tool_use"]
            if response.stop_reason != "tool_use" or not tool_uses:
                record["final_answer"] = text
                record["status"] = "answered" if response.stop_reason == "end_turn" else f"stopped ({response.stop_reason})"
                break

            tool_results = []
            for block in tool_uses:
                call = await execute_tool_call(session, block.name, block.input, turn)
                record["tool_calls"].append(call)
                content = call["result"] if isinstance(call["result"], str) else json.dumps(call["result"], ensure_ascii=False)
                tool_results.append({"type": "tool_result", "tool_use_id": block.id,
                                     "content": content, "is_error": call["is_error"]})
            messages.append({"role": "user", "content": tool_results})
        else:
            record["status"] = f"gave up: all {max_turns} allowed model call(s) used without a final answer"
    record["finished_utc"] = datetime.now(timezone.utc).isoformat(timespec="seconds")
    return record
````

````python
def print_tool_call(call: dict) -> None:
    """Print one logged tool call: tool name, input arguments and the full returned result."""
    status = "ERROR" if call["is_error"] else "ok"
    print(f"[turn {call['turn']}] TOOL CALL   {call['tool']}({json.dumps(call['arguments'], ensure_ascii=False)})")
    print(f"[turn {call['turn']}] TOOL RESULT ({status}, {call['seconds']:.1f} s):")
    print(textwrap.indent(to_json(call["result"]), "    "))


def print_run_trace(run: dict | None) -> None:
    """Print the full reasoning chain of a run: the model's text and every tool call, turn by turn."""
    if run is None:
        print("(no run available)")
        return
    print(f"QUESTION: {run['question']}")
    print(f"MODEL: {run['model']} | status: {run['status']} | model calls: {run['llm_calls']} | "
          f"tool calls: {len(run['tool_calls'])} | tokens in/out: {run['input_tokens']}/{run['output_tokens']}")
    if run["server_env"]:
        print(f"SERVER ENV OVERRIDES: {run['server_env']}")
    texts = {entry["turn"]: entry["text"] for entry in run["assistant_text"]}
    last_turn = max([call["turn"] for call in run["tool_calls"]] + list(texts) + [0])
    for turn in range(1, last_turn + 1):
        show_text = turn in texts and not (run["final_answer"] and texts[turn] == run["final_answer"])
        calls = [c for c in run["tool_calls"] if c["turn"] == turn]
        if not (show_text or calls):
            continue
        print("-" * 100)
        if show_text:
            print(f"[turn {turn}] CLAUDE: " + textwrap.fill(texts[turn], 100, subsequent_indent="    "))
        for call in calls:
            print_tool_call(call)
    print("-" * 100)
    print("FINAL ANSWER: see the next cell" if run["final_answer"] else "FINAL ANSWER: none")


def show_final_answer(run: dict | None) -> None:
    """Render the final answer of a run as Markdown."""
    if run and run["final_answer"]:
        display(Markdown(f"**Final answer ({run['model']}):**\n\n" + run["final_answer"]))
    else:
        print("No final answer:", run["status"] if run else "run not available")


def settings_of(run: dict) -> tuple:
    """Everything that must match for a cached run to be reused."""
    return (run["question"], run["model"], run["max_turns"], run["system_prompt"], run["server_env"])


def api_error_in(exc: BaseException) -> anthropic.APIError | None:
    """The Anthropic API error behind an exception, if any (anyio task groups wrap errors in ExceptionGroups)."""
    if isinstance(exc, anthropic.APIError):
        return exc
    if isinstance(exc, BaseExceptionGroup):
        return next((found for inner in exc.exceptions if (found := api_error_in(inner)) is not None), None)
    return None


async def get_run(label: str, question: str, *, max_turns: int = MAX_AGENT_TURNS,
                  server_env: dict[str, str] | None = None) -> dict | None:
    """Reload a cached run with identical settings, or execute the agent and cache the new run."""
    path = RUNS_DIR / f"{label}.json"
    wanted = (question, LLM_MODEL, max_turns, SYSTEM_PROMPT, server_env or {})
    if path.exists() and not FORCE_RERUN:
        cached = json.loads(path.read_text(encoding="utf-8"))
        if settings_of(cached) == wanted:
            print(f"Replaying cached run agent_runs/{path.name}, recorded {cached['started_utc']} "
                  "(set FORCE_RERUN = True to call the API again)\n")
            return cached
        print(f"Cached run {path.name} has different settings; running the agent again.")
    if not api_key_available:
        print(f"No cached run '{label}' and no ANTHROPIC_API_KEY: add your key to .env and re-run this cell.")
        return None
    try:
        run = await run_agent(question, max_turns=max_turns, server_env=server_env)
    except Exception as exc:               # e.g. invalid key, rate limit or overload that outlasted the retries
        api_error = api_error_in(exc)
        if api_error is None:
            raise
        print(f"Anthropic API error, run '{label}' not recorded: {type(api_error).__name__}: {api_error}")
        return None
    path.write_text(json.dumps(run, indent=2, ensure_ascii=False), encoding="utf-8")
    print(f"Ran the agent live; full log saved to agent_runs/{path.name}\n")
    return run


def successful_results(run: dict | None, tool: str) -> list:
    """Results of every successful call to one tool in a run (what the LLM actually saw)."""
    return [call["result"] for call in (run or {}).get("tool_calls", []) if call["tool"] == tool and not call["is_error"]]


def chain_check(run: dict | None, n_papers: int = 3) -> pd.DataFrame:
    """Did the agent chain both tools with sensible arguments and produce an answer?"""
    calls = (run or {}).get("tool_calls", [])
    simbad_ok = [c for c in calls if c["tool"] == "query_simbad" and not c["is_error"]]
    arxiv_ok = [c for c in calls if c["tool"] == "search_arxiv" and not c["is_error"]]
    errors = [c for c in calls if c["is_error"]]
    rows = [
        ("query_simbad succeeded", bool(simbad_ok),
         "; ".join(f"{c['arguments']['object_name']!r} -> {c['result']['main_id']!r}" for c in simbad_ok)),
        (f"search_arxiv succeeded with max_results = {n_papers}",
         any(c["arguments"].get("max_results") == n_papers for c in arxiv_ok),
         "; ".join(json.dumps(c["arguments"]) for c in arxiv_ok)),
        (f"{n_papers} papers returned", any(len(c["result"]) == n_papers for c in arxiv_ok),
         ", ".join(str(len(c["result"])) for c in arxiv_ok)),
        ("final answer produced", bool(run and run["final_answer"]), (run or {}).get("status") or "no run"),
        ("no tool errors", bool(run) and not errors, f"{len(errors)} error(s)"),
    ]
    return pd.DataFrame(rows, columns=["check", "passed", "detail"])
````

## 5. Physical plausibility

### 5.1 SIMBAD tool output vs. the SIMBAD web interface

On 2026-10-13 (around 05:56 UTC) I opened the SIMBAD web pages of both objects in a browser ([Betelgeuse](https://simbad.cds.unistra.fr/simbad/sim-id?Ident=Betelgeuse), [Proxima Centauri](https://simbad.cds.unistra.fr/simbad/sim-id?Ident=Proxima+Centauri)) and copied the "Basic data" values:

| Field | Betelgeuse (web) | Proxima Centauri (web) |
|---|---|---|
| Main identifier | `* alf Ori` | `NAME Proxima Centauri` |
| Object type | Red Supergiant | High Proper Motion Star |
| Spectral type | M1-M2Ia-Iab (quality B; Keenan & McNeil 1989, 1989ApJS...71..245K) | M5.5Ve (quality C; 1991AJ....101..662B) |
| Parallax | 6.55 ± 0.83 mas (quality A; Hipparcos new reduction, 2007A&A...474..653V) | 768.0665 ± 0.0499 mas (quality A; Gaia EDR3, 2020yCat.1350....0G) |
| Radial velocity | +21.91 ± 0.51 km/s (quality A; 2005A&A...430..165F) | −20.578199 ± 0.004684 km/s (quality A; 2020AJ....160..120J) |

The next cell compares these values, field by field, with what `query_simbad` returned **inside the agent runs**. The comparison therefore covers exactly the data the model received, not a separate query.

````python
SIMBAD_WEB_REFERENCE = {     # copied by hand from the SIMBAD web interface on 2026-10-13
    FIRST_OBJECT: {"main_id": "* alf Ori", "object_type_description": "Red Supergiant",
                   "spectral_type": "M1-M2Ia-Iab", "parallax_mas": 6.55, "parallax_error_mas": 0.83,
                   "radial_velocity_km_s": 21.91, "radial_velocity_error_km_s": 0.51},
    SECOND_OBJECT: {"main_id": "NAME Proxima Centauri", "object_type_description": "High Proper Motion Star",
                    "spectral_type": "M5.5Ve", "parallax_mas": 768.0665, "parallax_error_mas": 0.0499,
                    "radial_velocity_km_s": -20.578199, "radial_velocity_error_km_s": 0.004684},
}
runs_by_object = {FIRST_OBJECT: betelgeuse_run, SECOND_OBJECT: second_run}


def simbad_record(run: dict | None, main_id: str | None = None) -> dict:
    """Last successful query_simbad result of a run, optionally only for one SIMBAD main identifier."""
    results = [result for result in successful_results(run, "query_simbad")
               if main_id is None or result.get("main_id") == main_id]
    return results[-1] if results else {}


def record_for(name: str) -> dict:
    """SIMBAD record the model received for one of the two objects (empty dict if none)."""
    run = runs_by_object[name]
    return simbad_record(run, SIMBAD_WEB_REFERENCE[name]["main_id"]) or simbad_record(run)


crosscheck_rows = []
for name, reference in SIMBAD_WEB_REFERENCE.items():
    record = record_for(name)
    for field, web_value in reference.items():
        tool_value = record.get(field)
        if isinstance(web_value, float):
            agrees = tool_value is not None and math.isclose(tool_value, web_value, rel_tol=1e-6)
        else:
            agrees = tool_value == web_value
        crosscheck_rows.append({"object": name, "field": field, "SIMBAD web page": web_value,
                                "query_simbad in agent run": tool_value, "agree": agrees})

simbad_crosscheck = pd.DataFrame(crosscheck_rows)
simbad_crosscheck_passed = bool(simbad_crosscheck["agree"].all())
print(f"{simbad_crosscheck['agree'].sum()} of {len(simbad_crosscheck)} fields agree with the web interface")
simbad_crosscheck
````

**Did the final answers pass the catalogue values on faithfully?** Agreement between the tool and the web page does not guarantee that the model quoted the tool correctly, so the next cell searches each final answer for the spectral type, the parallax and its uncertainty, either verbatim or correctly rounded.

````python
# ---- Text utilities shared by every check of the final answers (Sections 5 and 6) ----
LATEX_REPLACEMENTS = {r"\pm": "±", r"\approx": "≈", r"\sim": "~", r"\odot": "⊙", r"\star": "★",
                      r"\varpi": "parallax ", r"\times": "×", r"\,": " ", r"\;": " ", r"\!": "", r"\ ": " "}
NEW_STYLE_ARXIV_ID = r"(?<![\d.])\d{4}\.\d{4,5}(?:v\d+)?"              # e.g. 2608.19339v2
OLD_STYLE_ARXIV_ID = r"\b[a-z][a-z-]*(?:\.[A-Z]{2})?/\d{7}(?:v\d+)?"     # e.g. astro-ph/0601234v1
ARXIV_ID_PATTERN = re.compile(f"{NEW_STYLE_ARXIV_ID}|{OLD_STYLE_ARXIV_ID}")
LUMINOSITY = r"(?:Iab|Ia|Ib|III|II|IV|V)"
SPECTRAL_TYPE_PATTERN = re.compile(rf"\b[OBAFGKM]\d(?:\.\d)?(?:\s?[-/]\s?[OBAFGKM]?\d(?:\.\d)?)?"
                                   rf"(?:\s?{LUMINOSITY}(?:\s?[-/]\s?{LUMINOSITY})?[a-z]*|(?!\w|\.\d))")
QUANTITY_PATTERN = re.compile(
    r"(?<![\w.])(?P<number>-?\d{1,3}(?:,\d{3})+(?:\.\d+)?|-?\d+(?:\.\d+)?(?:[eE][-+]?\d+)?)"
    r"(?P<uncertainty>\s*(?:±|\+/-|\+-)\s*\d+(?:\.\d+)?|\s*\([^()]{1,20}\))?"
    r"(?P<unit>\s*(?:mas\b|milli-?arc-?sec\w*|parsecs?\b|pc\b|light[- ]years?|ly\b|km ?/ ?s\b|M_?⊙|M☉|M_?sun\b|"
    r"solar mass(?:es)?|R_?⊙|R☉|R_?sun\b|solar radi(?:us|i)|R_?★|AU\b|au\b|days?\b|d\b|Myr\b|Gyr\b|kyr\b|"
    r"yrs?\b|years?\b|mm\b|cm\b|µm\b|m\b|GHz\b|K\b|%))?")
UNIT_KEYS = [(r"mas|milli", "mas"), (r"parsec|pc", "pc"), (r"light|ly", "ly"), (r"km", "km/s"),
             (r"M_?⊙|M☉|M_?sun|solar mass", "M_sun"), (r"R_?⊙|R☉|R_?sun|solar radi", "R_sun"), (r"R_?★", "R_star"),
             (r"AU|au", "AU"), (r"day|d$", "d"), (r"Myr", "Myr"), (r"Gyr", "Gyr"), (r"kyr", "kyr"), (r"yr|year", "yr"),
             (r"mm", "mm"), (r"cm", "cm"), (r"µm", "µm"), (r"m$", "m"), (r"GHz", "GHz"), (r"K$", "K"), (r"%", "%")]


def plain(text: str) -> str:
    """Text with LaTeX markup, Markdown emphasis and typographic dashes reduced to plain characters."""
    for latex, character in LATEX_REPLACEMENTS.items():
        text = text.replace(latex, character)
    return re.sub(r"[$*`{}]", "", text).replace("−", "-").replace("–", "-").replace("—", "-")


def normalised(text: str) -> str:
    """Plain, lower-case text without whitespace (for comparing designations such as 'M1-M2 Ia-Iab')."""
    return re.sub(r"\s+", "", plain(text)).lower()


def blank_arxiv_ids(text: str) -> str:
    """Replace arXiv IDs (identifiers, not quantities) by spaces, keeping every other character in place."""
    return ARXIV_ID_PATTERN.sub(lambda match: " " * len(match.group()), text)


def unit_key(unit: str) -> str | None:
    """Canonical name of a unit as written after a number ('mas', 'pc', 'M_sun', ...), or None."""
    unit = unit.strip()
    return next((key for pattern, key in UNIT_KEYS if unit and re.match(pattern, unit)), None)


def quantities(text: str) -> list[dict]:
    """Numbers in a text with the unit written after them (an uncertainty may sit in between); arXiv IDs are skipped.
    Positions refer to blank_arxiv_ids(plain(text))."""
    found = []
    for match in QUANTITY_PATTERN.finditer(blank_arxiv_ids(plain(text))):
        number = match.group("number").replace(",", "")
        found.append({"text": number, "value": float(number), "unit": unit_key(match.group("unit") or ""),
                      "start": match.start(), "end": match.end()})
    return found


def number_supported(number: str, evidence: list[float]) -> bool:
    """True if a quoted number equals one of the evidence values, allowing for rounding."""
    value = abs(float(number))
    if "e" in number.lower():                                  # scientific notation: compare to 1 %
        return any(math.isclose(abs(e), value, rel_tol=0.01) for e in evidence)
    decimals = len(number.split(".")[1]) if "." in number else 0
    return any(abs(round(abs(e), decimals) - value) < 0.5 * 10**-decimals + 1e-12 for e in evidence)


def canonical_spectral_type(spectral_type: str) -> str:
    """Comparable form of a spectral type: 'M1–M2 Ia–Iab' -> 'm1-2ia-iab', 'M5.5 Ve' -> 'm5.5v'."""
    text = re.sub(r"\s+", "", plain(spectral_type))
    text = re.sub(r"([OBAFGKM])(\d(?:\.\d)?)-\1(\d)", r"\1\2-\3", text)       # M1-M2 -> M1-2
    text = re.sub(r"(?<=[IV])[c-z]+$", "", text)                              # peculiarity flags, e.g. e = emission
    return text.lower()


def spectral_types_in(text: str) -> list[str]:
    """Spectral-type designations (e.g. 'M1-M2Ia-Iab', 'M2 Iab', 'M5.5Ve') found in a text."""
    return [match.group().strip() for match in SPECTRAL_TYPE_PATTERN.finditer(plain(text))]


def parallax_windows(text: str) -> list[str]:
    """The rest of the sentence (at most 160 characters) after each mention of a parallax, where its value is quoted."""
    text = plain(text)
    windows = []
    for match in re.finditer(r"parallax|ϖ", text, flags=re.IGNORECASE):
        window = text[match.end(): match.end() + 160]
        windows.append(re.split(r"(?<!\d)\.(?!\d)|\n", window)[0])      # stop at the end of the sentence or line
    return windows


def how_quoted(run: dict | None, value) -> str:
    """How a catalogue value appears in the final answer: verbatim, equivalent/rounded, or not quoted."""
    answer = (run or {}).get("final_answer") or ""
    if isinstance(value, str):                                 # spectral type
        if normalised(value) in normalised(answer):
            return "verbatim"
        same = [s for s in spectral_types_in(answer) if canonical_spectral_type(s) == canonical_spectral_type(value)]
        return f"equivalent notation ({same[0]})" if same else "not quoted"
    # parallax and its error: only numbers written after the word "parallax" count, and only roundings that keep
    # the value to better than 15 % (so the "1" of "M1" can never pass for 0.83)
    numbers = [n for window in parallax_windows(answer) for n in re.findall(r"\d+(?:\.\d+)?", window)]
    if str(value) in numbers:
        return "verbatim"
    for number in numbers:
        decimals = len(number.split(".")[1]) if "." in number else 0
        step = 0.5 * 10**-decimals
        if abs(float(number) - abs(value)) <= step + 1e-12 and step <= 0.15 * abs(value):
            return f"rounded ({number})"
    return "not quoted"


faithfulness = pd.DataFrame([
    {"object": name, "catalogue value": f"{field} = {SIMBAD_WEB_REFERENCE[name][field]}",
     "in final answer": how_quoted(run, SIMBAD_WEB_REFERENCE[name][field])}
    for name, run in runs_by_object.items()
    for field in ("spectral_type", "parallax_mas", "parallax_error_mas")
])
faithfulness
````

### 5.2 Sanity checks on physical quantities

Agreement with SIMBAD only shows that the pipeline copied the catalogue correctly. The checks below ask whether the numbers make physical sense:

1. **Distance from parallax:** $d = 1000/\varpi$ pc, with first-order uncertainty $\sigma_d = 1000\,\sigma_\varpi/\varpi^2$. For Betelgeuse I compare with two independent estimates: $222^{+48}_{-34}$ pc from radio (VLA/ALMA/e-MERLIN) positions combined with Hipparcos (Harper et al. 2017, AJ 154, 11), and $168^{+27}_{-15}$ pc from seismic and evolutionary modelling (Joyce et al. 2020, ApJ 902, 63). Agreement within $2\sigma$ passes. Proxima must come out closer than α Cen AB (≈1.34 pc), because it is the nearest known star.
2. **Fractional parallax error:** above about 10 %, $1/\varpi$ becomes a biased distance estimator (Lutz–Kelker-type bias). This is flagged as a caution, not a failure.
3. **Radius from angular diameter:** $R = (\theta/2)\,d$. Interferometric diameters of Betelgeuse are $\theta_{LD} = 42.28 \pm 0.43$ mas (VLTI/AMBER, K-band continuum; Montargès et al. 2014, A&A 572, A17) and 47 mas (Michelson & Pease 1921, quoted in one of the retrieved abstracts). A red supergiant should have $R$ of several hundred to about a thousand $R_\odot$ (Joyce et al. 2020: $764^{+116}_{-62}\,R_\odot$), whereas a red giant has $R \sim 10$–$100\,R_\odot$.
4. **Spectral type vs. object type:** the luminosity class in the spectral type must agree with SIMBAD's independent object classification. Ia–Iab is a supergiant, so Betelgeuse should be classified as a red supergiant. V is a dwarf, so Proxima must not be classified as a giant or supergiant; its SIMBAD type, "High Proper Motion Star", says nothing about luminosity, so for Proxima this test can only rule out a contradiction.
5. **Radial velocity:** a star in the solar neighbourhood should have $|v_r| \lesssim 100$ km/s (disc velocity dispersions are tens of km/s). Thousands of km/s would signal a redshift/velocity mix-up.

````python
AU_IN_SOLAR_RADII = 215.032
PC_IN_LIGHT_YEARS = 3.26156
LUMINOSITY_CLASS_NAMES = {"Ia": "supergiant", "Iab": "supergiant", "Ib": "supergiant", "I": "supergiant",
                          "II": "bright giant", "III": "giant", "IV": "subgiant", "V": "dwarf (main sequence)"}
BETELGEUSE_DISTANCES_PC = {"Harper et al. 2017": (222.0, 48.0, 34.0), "Joyce et al. 2020": (168.0, 27.0, 15.0)}
BETELGEUSE_THETA_MAS = {"Montargès et al. 2014": 42.3, "Michelson & Pease 1921": 47.0}


def parallax_distance(parallax_mas: float, error_mas: float) -> tuple[float, float]:
    """Distance in pc and its first-order uncertainty from a parallax in mas."""
    return 1000.0 / parallax_mas, 1000.0 * error_mas / parallax_mas**2


def luminosity_class(spectral_type: str | None) -> str | None:
    """First MK luminosity class in a spectral type, e.g. 'M1-M2Ia-Iab' -> 'Ia', 'M5.5Ve' -> 'V', 'M2 Iab' -> 'Iab'."""
    match = re.search(r"[\d.]\s?(Iab|Ia|Ib|III|II|IV|V|I)", spectral_type or "")
    return match.group(1) if match else None


sanity_rows = []


def add_check(obj: str, check: str, value: str, expectation: str, verdict: str) -> None:
    sanity_rows.append({"object": obj, "check": check, "value": value, "expectation": expectation, "verdict": verdict})


for name in runs_by_object:
    record = record_for(name)
    if not record.get("parallax_mas") or record["parallax_mas"] <= 0 or record.get("parallax_error_mas") is None:
        add_check(name, "all checks", "no usable SIMBAD parallax in the run", "-", "n/a")
        continue
    distance, distance_err = parallax_distance(record["parallax_mas"], record["parallax_error_mas"])
    fractional_error = record["parallax_error_mas"] / record["parallax_mas"]
    add_check(name, "fractional parallax error", f"{100 * fractional_error:.3g} %", "< 10 % for an unbiased 1/parallax",
              "PASS" if fractional_error < 0.10 else "CAUTION")

    if name == FIRST_OBJECT:
        for source, (lit, upper, lower) in BETELGEUSE_DISTANCES_PC.items():
            sigma = math.hypot(distance_err, lower if distance < lit else upper)
            z = abs(distance - lit) / sigma
            add_check(name, f"distance vs. {source}", f"{distance:.1f} ± {distance_err:.1f} pc",
                      f"{lit:.0f} (+{upper:.0f}/-{lower:.0f}) pc; |Δ| = {z:.1f}σ", "PASS" if z < 2 else "FAIL")
        for source, theta in BETELGEUSE_THETA_MAS.items():
            radius = (theta / 2 / 1000) * distance * AU_IN_SOLAR_RADII      # (arcsec x pc) = AU
            add_check(name, f"radius from θ = {theta} mas ({source})", f"{radius:.0f} R_sun",
                      "500-1200 R_sun for a red supergiant", "PASS" if 500 <= radius <= 1200 else "FAIL")
    else:
        add_check(name, "nearest-star test", f"{distance:.5f} pc = {distance * PC_IN_LIGHT_YEARS:.4f} ly",
                  "< 1.34 pc (closer than α Cen AB)", "PASS" if distance < 1.34 else "FAIL")

    lum_class = luminosity_class(record.get("spectral_type"))
    class_name = LUMINOSITY_CLASS_NAMES.get(lum_class, "unknown")
    object_type = (record.get("object_type_description") or "").lower()
    if lum_class is None:
        consistent, expectation = None, "a luminosity class in the spectral type"
    elif class_name == "supergiant":
        consistent, expectation = "supergiant" in object_type, "object type is a supergiant"
    else:
        consistent, expectation = "giant" not in object_type, "not classified as giant/supergiant (rules out a contradiction only)"
    add_check(name, "luminosity class vs. object type",
              f"{record.get('spectral_type')} -> class {lum_class} ({class_name}); type '{record.get('object_type_description')}'",
              expectation, "n/a" if consistent is None else "PASS" if consistent else "FAIL")

    velocity = record.get("radial_velocity_km_s")
    add_check(name, "radial velocity", f"{velocity:+.2f} km/s" if velocity is not None else "null",
              "|v_r| < 100 km/s for a nearby disc star",
              "n/a" if velocity is None else "PASS" if abs(velocity) < 100 else "FAIL")

sanity_checks = pd.DataFrame(sanity_rows)
sanity_failures = int((sanity_checks["verdict"] == "FAIL").sum())
print(f"Sanity checks: {len(sanity_checks)} run, {sanity_failures} failed, "
      f"{int((sanity_checks['verdict'] == 'CAUTION').sum())} caution")
sanity_checks
````

**Interpretation (using the catalogue values in the table of Section 5.1).** Betelgeuse's Hipparcos parallax gives $d = 1000/6.55 \approx 153 \pm 19$ pc. This agrees with the seismic estimate of Joyce et al. (168 pc, $0.6\sigma$) and lies $1.8\sigma$ below the radio–Hipparcos solution of Harper et al. (222 pc), so both pass the $2\sigma$ test. The 12.7 % fractional error is exactly the regime in which $1/\varpi$ is biased. Betelgeuse's angular diameter (≈42–47 mas) is also about seven times larger than its parallax, so convective motion of the photocentre adds astrometric noise. The spread among published distances (roughly 150–220 pc) is therefore expected, and an LLM quoting "about 150 pc" from this parallax without caveats would overstate the precision. With this distance, the measured angular diameter gives $R \approx 690$–$770\,R_\odot$, squarely in the red-supergiant range and consistent with luminosity class Ia–Iab, so the catalogue entries are physically coherent. Proxima's Gaia parallax gives $d = 1.3020$ pc $= 4.2465$ ly, the textbook value. It is closer than α Cen AB and has a relative error of $6.5 \times 10^{-5}$. Its class V spectral type (M5.5Ve, "e" for emission lines from chromospheric activity) identifies a low-mass main-sequence star. Both radial velocities, +21.9 and −20.6 km/s, are typical of nearby disc stars.

### 5.3 The LLM's summary vs. the raw abstracts

**Purpose:** test whether every statement the model made about the papers is supported by the abstracts it actually received. Four complementary checks:

1. **Citation check (deterministic):** every arXiv ID cited in the answer must be one that `search_arxiv` returned, and all three returned papers should be discussed.
2. **Number check (deterministic):** every number in the answer should occur, possibly rounded and with the same unit, in the physical content of the tool results, i.e. the SIMBAD values or the text of a title or abstract (so "15 M☉" is not matched by a "15 mm" baseline). A number without a unit may also match a date or reference. Other numbers are flagged: they are either derived by the model (e.g. a distance computed from the parallax) or recalled from memory.
3. **Evolutionary-claim check (deterministic):** does the answer give the evolutionary class implied by the SIMBAD luminosity class (supergiant for Betelgeuse, main-sequence dwarf for Proxima)? Which evolution-related statements in the answer (burning stage, supernova, Great Dimming, ages, masses, ...) have no counterpart in any abstract or SIMBAD field? Only affirmative clauses count, so "unlike Betelgeuse, Proxima will never explode as a supernova" is not read as a supernova claim, and quoted masses are compared number by number.
4. **Claim-by-claim check (LLM-assisted):** a separate Claude call without tools receives the final answer and the raw tool outputs. It labels each factual claim as SUPPORTED, NOT_IN_SOURCES (possibly true, but not stated in the abstracts or the SIMBAD record) or CONTRADICTED. The same model family is checking itself, so these verdicts are an aid only; checks 1–3 do not depend on the model.

First, the raw abstracts the model received:

````python
def base_arxiv_id(arxiv_id: str) -> str:
    """arXiv identifier without its version suffix, e.g. '2608.19339v2' -> '2608.19339'."""
    return re.sub(r"v\d+$", "", arxiv_id)


def arxiv_ids_in(text: str) -> set[str]:
    """Version-free arXiv identifiers mentioned in a text (plain IDs, 'arXiv:' prefixes and abs/ URLs)."""
    return {base_arxiv_id(match.group()) for match in ARXIV_ID_PATTERN.finditer(plain(text))}


def returned_papers(run: dict | None) -> list[dict]:
    """Papers from every successful search_arxiv call of a run, de-duplicated by arXiv ID."""
    papers = {}
    for result in successful_results(run, "search_arxiv"):
        for paper in result:
            papers.setdefault(base_arxiv_id(paper["arxiv_id"]), paper)
    return list(papers.values())


for name, run in runs_by_object.items():
    print("=" * 100)
    print(f"{name}: raw abstracts returned by search_arxiv")
    for paper in returned_papers(run):
        print("-" * 100)
        print(f"arXiv:{paper['arxiv_id']} ({paper['published']}, {paper['primary_category']}) {paper['title']}")
        print(f"Authors: {', '.join(paper['authors'][:6])}{' et al.' if len(paper['authors']) > 6 else ''}")
        print(textwrap.fill(paper["abstract"], 100))
````

**What these abstracts contain (snapshot of 2026-10-13).** When I prepared this analysis, arXiv returned the following three most recent matches for each object. The next cell checks whether the run analysed here returned the same papers.

*Betelgeuse*
* **2609.27055** (Friedman & Goman, physics education): a two-hole masked-telescope lab that measures the angular diameter of Venus. Betelgeuse appears only as the target of Michelson & Pease's 1921 measurement (47 mas with a 6.1 m baseline). The paper says nothing about Betelgeuse's evolution.
* **2608.19339** (Dent et al., ALMA): calls Betelgeuse "one of the closest RSGs". It finds a ≈2300 K sub-mm photosphere at 1.1–1.3 $R_\star$ with hot patches (≈+800 K) that have persisted since 2015, thought to be associated with shocks driven by underlying convective cells, and no clear evidence of rotation.
* **2608.11672** (Goldberg et al.): argues that the reported companion α Ori B is seen in Betelgeuse's light reflected by the companion's bow shock. This reconciles the detections with HST far-UV non-detections and disfavours a hot 3 $M_\odot$ companion.

The only evolutionary statement these abstracts support is that Betelgeuse is a **red supergiant** with a convective, extended, mass-losing atmosphere and possibly a low-mass companion. Claims about its nuclear-burning stage (core helium or carbon burning), its age or mass, the time until core collapse, or the 2019–2020 Great Dimming cannot come from these sources. If the summary contains them, the checks below should flag them.

*Proxima Centauri*
* **2609.05599** (Doshi et al., POSTELLAR): a diffusion-model method for sampling stellar spectra. Proxima's empirical spectrum is used only to build synthetic SPIRou test data.
* **2608.12686** (Azizi et al.): a TiO 567 nm index measured in HARPS spectra reveals an activity cycle of $2873^{+47}_{-54}$ days (≈7.9 yr).
* **2608.10212** (Males et al., GMagAO-X): Proxima Centauri b is named as a driving science case for the GMT's high-contrast imager.

Only the activity-cycle paper concerns the star itself, and none of the three states its evolutionary stage. "Main-sequence M dwarf" follows from the SIMBAD spectral type (luminosity class V), not from the papers. Statements about its age, its main-sequence lifetime of trillions of years or its being fully convective are background knowledge. The search also illustrates failure mode F4 (Section 6): two of the three "most recent" papers mention Proxima only in passing.

````python
SNAPSHOT_2026_10_13 = {FIRST_OBJECT: ["2609.27055", "2608.19339", "2608.11672"],
                       SECOND_OBJECT: ["2609.05599", "2608.12686", "2608.10212"]}
snapshot_matches = {}
for name, run in runs_by_object.items():
    ids = [base_arxiv_id(paper["arxiv_id"]) for paper in returned_papers(run)]
    snapshot_matches[name] = ids == SNAPSHOT_2026_10_13[name]
    note = ("same papers as the snapshot: the description above applies" if snapshot_matches[name]
            else "DIFFERENT papers from the snapshot (e.g. newer submissions): rely on the abstracts printed above")
    print(f"{name}: run returned {ids} -> {note}")
````

````python
def simbad_evidence(record: dict) -> str:
    """SIMBAD values written out with their units, so that quoted numbers can be matched unit by unit."""
    if not record:
        return ""
    return (f"object type: {record.get('object_type_description')}; spectral type: {record.get('spectral_type')}; "
            f"parallax {record.get('parallax_mas')} mas ± {record.get('parallax_error_mas')} mas; "
            f"radial velocity {record.get('radial_velocity_km_s')} km/s ± {record.get('radial_velocity_error_km_s')} km/s; "
            f"redshift {record.get('redshift')}")


def physical_evidence(run: dict | None) -> str:
    """Physical content of a run's successful tool outputs: SIMBAD values, paper titles and abstracts."""
    parts = []
    for call in (run or {}).get("tool_calls", []):
        if call["is_error"]:
            continue
        if call["tool"] == "query_simbad":
            parts.append(simbad_evidence(call["result"]))
        elif call["tool"] == "search_arxiv":
            parts += [f"{paper['title']}\n{paper['abstract']}" for paper in call["result"]]
    return "\n".join(parts)


def metadata_evidence(run: dict | None) -> str:
    """Dates, references and names in the tool outputs: they can support a quoted year, never a physical quantity."""
    parts = []
    for call in (run or {}).get("tool_calls", []):
        if call["is_error"]:
            continue
        if call["tool"] == "query_simbad":
            parts += [str(call["result"].get(field)) for field in ("main_id", "parallax_reference", "radial_velocity_reference")]
        elif call["tool"] == "search_arxiv":
            parts += [f"{paper['published']} {paper['updated']} {', '.join(paper['authors'])}" for paper in call["result"]]
    return "\n".join(parts)


def quantity_supported(quantity: dict, physical: list[dict], metadata: list[dict]) -> bool:
    """A quoted number counts as sourced if a tool output contains it (allowing rounding) with a compatible unit.
    A number with a unit needs a physical value with the same unit, or a unit-less number in the physical text;
    a number without a unit may also match a date or reference."""
    if quantity["unit"] is None:
        pool = [q["value"] for q in physical + metadata]
    else:
        pool = [q["value"] for q in physical if q["unit"] in (quantity["unit"], None)]
    return number_supported(quantity["text"], pool)


def citation_and_number_checks(run: dict | None) -> tuple[dict, pd.DataFrame]:
    """Deterministic grounding checks of a final answer against the tool outputs of the same run."""
    answer = (run or {}).get("final_answer") or ""
    returned_ids = {base_arxiv_id(paper["arxiv_id"]) for paper in returned_papers(run)}
    cited_ids = arxiv_ids_in(answer)
    citations = {"returned": sorted(returned_ids), "cited": sorted(cited_ids),
                 "cited but never returned": sorted(cited_ids - returned_ids),
                 "returned but not cited": sorted(returned_ids - cited_ids)}

    physical, metadata = quantities(physical_evidence(run)), quantities(metadata_evidence(run))
    text = blank_arxiv_ids(plain(answer))          # the text that quantities() positions refer to
    rows = []
    for quantity in quantities(answer):
        if quantity["unit"] is None and "." not in quantity["text"] and abs(quantity["value"]) < 10:
            continue                                # small counts ("3 papers") are not quantities
        if not quantity_supported(quantity, physical, metadata):
            context = text[max(quantity["start"] - 60, 0): quantity["end"] + 30]
            rows.append({"number": quantity["text"] + (f" {quantity['unit']}" if quantity["unit"] else ""),
                         "context in answer": "..." + " ".join(context.split()) + "..."})
    return citations, pd.DataFrame(rows, columns=["number", "context in answer"])


grounding = {}
for name, run in runs_by_object.items():
    citations, unsupported_numbers = citation_and_number_checks(run)
    grounding[name] = {"citations": citations, "unsupported_numbers": unsupported_numbers}
    print(f"===== {name} =====")
    for key, value in citations.items():
        print(f"  {key:26s}: {value}")
    print(f"  numbers not found (with a compatible unit) in the SIMBAD values or the abstracts: {len(unsupported_numbers)}")
    display(unsupported_numbers)
````

````python
NEGATION = re.compile(r"\b(?:not|never|no longer|unlike|rather than|instead of|neither|nor|cannot)\b|n't\b", re.IGNORECASE)
EXPECTED_STAGE_WORDING = {"supergiant": r"supergiant|\bRSG\b",
                          "dwarf (main sequence)": r"main[- ]sequence|\bdwarf\b|hydrogen[- ]burning"}
CONTRADICTING_WORDING = {"supergiant": r"main[- ]sequence star|\bred dwarf\b|\bred giant\b",
                         "dwarf (main sequence)": r"supergiant|\bred giant\b|giant branch|supernova"}
EVOLUTION_TERMS = [   # (statement, wording that reveals it, unit of its quantities to compare, if any)
    ("nuclear burning stage", r"burning|\bfusion\b|\bfus(?:e|es|ing)\b", None),
    ("supernova / core collapse", r"supernova|core[- ]collapse|\bexplo(?:de|des|sion)", None),
    ("Great Dimming (2019-2020)", r"dimming", None),
    ("main sequence", r"main[- ]sequence", None),
    ("red giant / giant branch", r"giant branch|\bred giant\b", None),
    ("red supergiant", r"supergiant|\bRSGs?\b", None),
    ("mass loss", r"mass[- ]loss|los(?:e|es|ing) mass|stellar winds?", None),
    ("age / lifetime", r"\b(?:age|ages|lifetime|lifespan)\b|years old|(?:million|billion|trillion)s? (?:of )?years|"
                       r"\b[MG]yrs?\b", None),
    ("stellar mass", r"solar mass|M_?⊙|M☉|M_?sun\b", "M_sun"),
    ("fully convective", r"fully convective", None),
    ("magnetic activity", r"activity cycle|\bflares?\b|magnetic activity|chromospher", None),
]


def clauses(text: str) -> list[str]:
    """Clauses of a text: sentences, bullet items and table cells, also split at commas, semicolons and 'but'."""
    parts = re.split(r"(?<!\d)[.,](?!\d)|[;:!?\n|]|\bbut\b|\bwhereas\b|\bwhile\b", plain(text))
    return [part.strip() for part in parts if part.strip()]


def affirmative_clauses(text: str) -> list[str]:
    """Clauses without a negation or contrast ('not', 'never', 'unlike', ...), i.e. statements actually made."""
    return [clause for clause in clauses(text) if not NEGATION.search(clause)]


def evolution_checks(name: str, run: dict | None, record: dict) -> tuple[dict, list[dict]]:
    """For one object: does the answer state the evolutionary class implied by the SIMBAD luminosity class, and which
    evolution-related statements in it have no counterpart in the abstracts or the SIMBAD record?"""
    statements = affirmative_clauses((run or {}).get("final_answer") or "")
    lum_class = luminosity_class(record.get("spectral_type"))
    class_name = LUMINOSITY_CLASS_NAMES.get(lum_class)
    # the luminosity class read from the SIMBAD spectral type counts as a source (e.g. class V -> main sequence)
    sources = plain(physical_evidence(run)) + f"\nluminosity class from the SIMBAD spectral type: {class_name}"
    source_masses = [q["value"] for q in quantities(sources) if q["unit"] == "M_sun"]
    expected, contradicting = EXPECTED_STAGE_WORDING.get(class_name), CONTRADICTING_WORDING.get(class_name)
    contradictions = sorted({m.group().lower() for clause in statements if contradicting
                             for m in re.finditer(contradicting, clause, flags=re.IGNORECASE)})
    stage_row = {"object": name, "SIMBAD luminosity class": f"{lum_class} ({class_name})",
                 "stated in final answer": (None if expected is None else       # None: no luminosity class to compare
                                            any(re.search(expected, c, flags=re.IGNORECASE) for c in statements)),
                 "contradicting wording (check context)": ", ".join(contradictions) or "none"}
    rows = []
    for term, pattern, unit in EVOLUTION_TERMS:
        hits = [clause for clause in statements if re.search(pattern, clause, flags=re.IGNORECASE)]
        if not hits:
            continue
        values = [q for clause in hits for q in quantities(clause) if unit and q["unit"] == unit]
        if values:            # compare the quoted numbers themselves (e.g. 18 M_sun vs. a 3 M_sun companion)
            in_sources = all(number_supported(q["text"], source_masses) for q in values)
        else:
            in_sources = bool(re.search(pattern, sources, flags=re.IGNORECASE))
        rows.append({"object": name, "statement in answer": term, "clause": hits[0][:120],
                     "found in abstracts/SIMBAD": in_sources})
    return stage_row, rows


stage_rows, term_rows = [], []
for name, run in runs_by_object.items():
    stage_row, rows = evolution_checks(name, run, record_for(name))
    stage_rows.append(stage_row)
    term_rows += rows

stage_check = pd.DataFrame(stage_rows)
evolution_terms = pd.DataFrame(term_rows, columns=["object", "statement in answer", "clause", "found in abstracts/SIMBAD"]
                               ).astype({"found in abstracts/SIMBAD": bool})
stage_consistent = bool(len(stage_check)) and all(value is True for value in stage_check["stated in final answer"])
display(stage_check)
display(evolution_terms)
````

````python
VERIFIER_INSTRUCTIONS = """You check an AI assistant's answer against the only sources it was given.

Split the ANSWER into its individual factual claims about the object or the papers (skip formatting, headings and
sentences that only restate the question). Judge each claim against the SOURCES only, never against your own
knowledge:
- "SUPPORTED": stated in, or directly implied by, the sources;
- "NOT_IN_SOURCES": not stated in the sources (it may be true background knowledge, but the sources do not support it);
- "CONTRADICTED": conflicts with the sources.
A value derived from the sources by simple arithmetic (e.g. a distance computed as 1000/parallax) counts as SUPPORTED.

Reply with JSON only, in exactly this form:
{"claims": [{"claim": "...", "verdict": "SUPPORTED|NOT_IN_SOURCES|CONTRADICTED",
             "source": "arXiv ID, SIMBAD or null", "explanation": "one short sentence"}]}"""


def verifier_prompt(run: dict, name: str) -> str:
    """Build the verifier input: SIMBAD record, raw abstracts and the answer to check."""
    abstracts = "\n\n".join(f"[arXiv:{p['arxiv_id']}] {p['title']}\n{p['abstract']}" for p in returned_papers(run))
    return (f"SOURCES\n<simbad_record>\n{json.dumps(record_for(name), indent=1)}\n</simbad_record>\n"
            f"<abstracts>\n{abstracts}\n</abstracts>\n\nANSWER\n<answer>\n{run['final_answer']}\n</answer>")


def parse_claims(text: str) -> list[dict] | None:
    """Claims list from the verifier's JSON reply (a Markdown fence is tolerated), or None if it is malformed."""
    try:
        claims = json.loads(text[text.find("{"): text.rfind("}") + 1])["claims"]
    except (ValueError, KeyError, TypeError):
        return None
    if not isinstance(claims, list):
        return None
    return [claim for claim in claims if isinstance(claim, dict) and {"claim", "verdict"} <= claim.keys()]


async def verify_claims(label: str, name: str) -> pd.DataFrame | None:
    """Claim-by-claim verification of a final answer; the verdicts are cached inside the run's JSON file."""
    run = runs_by_object[name]
    if not run or not run["final_answer"]:
        print(f"{label}: no final answer to verify")
        return None
    prompt = verifier_prompt(run, name)
    cached = run.get("verification") or {}
    if FORCE_RERUN or cached.get("claims") is None or cached.get("prompt") != prompt:
        if not api_key_available:
            print(f"{label}: no cached verification and no API key")
            return None
        try:
            async with anthropic.AsyncAnthropic(max_retries=6) as llm:
                reply = await llm.messages.create(model=LLM_MODEL, max_tokens=LLM_MAX_TOKENS, system=VERIFIER_INSTRUCTIONS,
                                                  messages=[{"role": "user", "content": prompt}])
        except Exception as exc:
            api_error = api_error_in(exc)
            if api_error is None:
                raise
            print(f"{label}: Anthropic API error, verification skipped ({type(api_error).__name__}: {api_error})")
            return None
        reply_text = "".join(block.text for block in reply.content if block.type == "text")
        run["verification"] = {"model": LLM_MODEL, "prompt": prompt, "claims": parse_claims(reply_text),
                               "raw_reply": reply_text}
        (RUNS_DIR / f"{label}.json").write_text(json.dumps(run, indent=2, ensure_ascii=False), encoding="utf-8")
    if run["verification"]["claims"] is None:
        print(f"{label}: the verifier's reply was not valid JSON (stored as raw_reply in agent_runs/{label}.json); "
              "re-run this cell to try again")
        return None
    table = pd.DataFrame(run["verification"]["claims"], columns=["claim", "verdict", "source", "explanation"])
    table["verdict"] = table["verdict"].astype(str).str.strip().str.upper().str.replace(" ", "_")
    return table


claim_tables = {}
for label, name in [("betelgeuse", FIRST_OBJECT), ("proxima_centauri", SECOND_OBJECT)]:
    claim_tables[name] = await verify_claims(label, name)
    table = claim_tables[name]
    if table is not None:
        print(f"===== {name}: {len(table)} claims -> {table['verdict'].value_counts().to_dict()}")
        display(table)
````

````python
def discrepancy_report() -> str:
    """Markdown summary of every flagged discrepancy between the final answers and the raw tool outputs."""
    lines = []
    for name in runs_by_object:
        lines.append(f"**{name}**")
        citations = grounding[name]["citations"]
        lines.append(f"* Citations: {len(citations['cited'])} arXiv IDs cited; cited but never returned by the tool: "
                     f"{', '.join(citations['cited but never returned']) or 'none'}; returned but not cited: "
                     f"{', '.join(citations['returned but not cited']) or 'none'}.")
        numbers = grounding[name]["unsupported_numbers"]
        lines.append("* Numbers not found in the SIMBAD values or the abstracts: "
                     + (", ".join(f"`{n}`" for n in numbers["number"]) if len(numbers) else "none") + ".")
        stage = stage_check[stage_check["object"] == name].iloc[0]
        stated = {True: "stated", False: "**not stated**", None: "not assessable (no SIMBAD luminosity class)"}[
            stage["stated in final answer"]]
        lines.append(f"* Evolutionary class from the luminosity class ({stage['SIMBAD luminosity class']}): "
                     f"{stated} in the answer; "
                     f"contradicting wording: {stage['contradicting wording (check context)']}.")
        unsupported_terms = evolution_terms[(evolution_terms["object"] == name) & ~evolution_terms["found in abstracts/SIMBAD"]]
        lines.append("* Evolution-related statements with no counterpart in the abstracts or SIMBAD: "
                     + (", ".join(unsupported_terms["statement in answer"]) if len(unsupported_terms) else "none") + ".")
        table = claim_tables.get(name)
        if table is None:
            lines.append("* Claim-by-claim check: not available (no answer, no API key or an invalid verifier reply).")
        else:
            flagged = table[table["verdict"] != "SUPPORTED"]
            lines.append(f"* Claim-by-claim check: {len(table)} claims, {int((table['verdict'] == 'SUPPORTED').sum())} "
                         f"supported, {len(flagged)} flagged:")
            lines += [f"    * **{row.verdict}** - \"{row.claim}\" ({row.explanation})" for row in flagged.itertuples()]
        lines.append("")
    return "\n".join(lines)


display(Markdown("#### Flagged discrepancies between the summaries and the raw abstracts\n\n" + discrepancy_report()))
````

**How to read the flags.** A statement with no counterpart in the sources is not necessarily a factual error: the model often adds correct textbook knowledge, for example that Betelgeuse is expected to end as a core-collapse supernova. Such additions still count as discrepancies for this assignment, because the question asked what *the papers* suggest, and the answer presents background knowledge as if the abstracts had said it. The most important discrepancies are (i) evolutionary-stage statements attributed to papers that do not discuss evolution (see the snapshot above), (ii) numbers that appear in no tool output, (iii) an evolutionary class that contradicts the SIMBAD luminosity class, and (iv) any CONTRADICTED claim, which would be an outright misreading of a source.

## 6. Deliberately triggered failure modes

**Purpose:** document where and how the pipeline fails. The four cases cover different layers:

| Case | Layer | Trigger | Deterministic? |
|---|---|---|---|
| F1 | Tool input (entity resolution) | misspelled identifier `"Betelguese"` | trigger yes; recovery depends on the model |
| F2 | Upstream service | SIMBAD unreachable, simulated by starting the server with `SIMBAD_TIMEOUT_S=0.001` | trigger yes; the model's response varies |
| F3 | Agent control loop | turn budget of 1 model call (`max_turns=1`) | yes |
| F4 | Retrieval semantics (silent) | "most recent papers whose abstracts mention X" returns papers that mention X only in passing | depends on arXiv's current listings |

Each cell runs the trigger, prints the full trace, decides automatically whether the loop recovered, and writes the observation into the matching `<!-- BEGIN AUTO Fx -->` block of `week05_failures_ly.md`. The written-out analysis and the proposed production fixes are in that file.

````python
def update_failure_doc(key: str, body: str) -> None:
    """Replace the auto-generated block of one failure case in week05_failures_ly.md and display it."""
    start, end = f"<!-- BEGIN AUTO {key} -->", f"<!-- END AUTO {key} -->"
    document = FAILURES_MD.read_text(encoding="utf-8")
    pattern = re.compile(re.escape(start) + r".*?" + re.escape(end), flags=re.DOTALL)
    assert pattern.search(document), f"markers for {key} not found in {FAILURES_MD.name}"
    block = f"{start}\n{body.strip()}\n{end}"
    FAILURES_MD.write_text(pattern.sub(lambda _: block, document), encoding="utf-8")
    display(Markdown(body))


def quote(text: str | None, limit: int = 700) -> str:
    """Format (part of) a text as a Markdown block quote."""
    text = (text or "(none)").strip()
    text = text if len(text) <= limit else text[:limit].rstrip() + " [...]"
    return "\n".join("> " + line for line in text.splitlines())


def call_summary(run: dict) -> str:
    """One Markdown bullet per tool call: arguments and outcome."""
    lines = []
    for call in run["tool_calls"]:
        outcome = "**error**: " + call["result"].replace("\n", " ")[:220] if call["is_error"] else "ok"
        lines.append(f"    * turn {call['turn']}: `{call['tool']}({json.dumps(call['arguments'])})` -> {outcome}")
    return "\n".join(lines) or "    * (no tool calls)"


def observation_header(run: dict) -> str:
    return (f"*Observed in `week05_ly.ipynb` (run recorded {run['started_utc']}, model `{run['model']}`, "
            f"status: {run['status']}, {run['llm_calls']} model call(s)).*")


failure_outcomes = {}
````

### F1 — Misspelled object name

The user supplies the identifier `"Betelguese"`, a common misspelling. SIMBAD's identifier matching is exact, so `query_simbad` returns an error. The question is whether the model notices the typo and retries.

````python
def correction_acknowledged(answer: str) -> bool:
    """True if the answer makes a changed identifier visible: it repeats the user's spelling or mentions the failed lookup."""
    return bool(re.search(r"betelguese|misspel|spelling|\btypo\b|did you mean|correct (?:name|spelling|identifier)|"
                          r"no (?:match|object|results?)|not found|could ?n[o']t find|unable to find",
                          plain(answer), flags=re.IGNORECASE))


f1_question = "Query SIMBAD for the object 'Betelguese' and report its spectral type and parallax."
f1_run = await get_run("failure_F1_misspelled_name", f1_question)
print_run_trace(f1_run)

if f1_run:
    simbad_calls = [c for c in f1_run["tool_calls"] if c["tool"] == "query_simbad"]
    errors = [c for c in simbad_calls if c["is_error"]]
    successes = [c for c in simbad_calls if not c["is_error"]]
    answer = f1_run["final_answer"] or ""
    acknowledged = correction_acknowledged(answer)
    if not simbad_calls:
        verdict = ("**Not triggered:** the model never called `query_simbad`, so the answer contains no SIMBAD data; "
                   "manual intervention would be needed to obtain it.")
        failure_outcomes["F1"] = "not triggered (no SIMBAD call)"
    elif not errors:
        verdict = ("**Not triggered at the tool level:** the model changed the spelling before its first call "
                   f"({', '.join(repr(c['arguments']['object_name']) for c in simbad_calls)}), i.e. it rewrote the user's input "
                   + ("and the answer mentions the original spelling or the correction." if acknowledged else
                      "and the answer mentions neither the user's spelling nor the correction."))
        failure_outcomes["F1"] = "masked by the model"
    elif successes:
        verdict = (f"**Recovered without manual intervention.** After {len(errors)} failed call(s) the model retried with "
                   f"`{successes[0]['arguments']['object_name']}`, which SIMBAD resolved to `{successes[0]['result']['main_id']}`. "
                   + ("The answer mentions the user's spelling or the failed lookup, so the change of identifier is "
                      "visible." if acknowledged else "The answer mentions neither the user's spelling nor the failed "
                      "lookup, so the change of identifier is invisible to the user."))
        failure_outcomes["F1"] = "recovered automatically"
    else:
        verdict = (f"**Did not recover; manual intervention required.** All {len(errors)} `query_simbad` call(s) failed, so "
                   "the answer contains no catalogue data from SIMBAD.")
        failure_outcomes["F1"] = "manual intervention required"
    update_failure_doc("F1", "\n".join([
        observation_header(f1_run), "",
        "* **Tool calls:**", call_summary(f1_run),
        f"* **Recovery:** {verdict}",
        "* **Final answer (excerpt):**", "", quote(answer)]))
````

### F2 — Upstream service unavailable

The server is started with `SIMBAD_TIMEOUT_S=0.001`, so every SIMBAD request is abandoned after 1 ms: from the agent's point of view SIMBAD behaves like a service that is down. arXiv still works. The full Betelgeuse question is asked. A robust agent should answer the arXiv part and state clearly that the catalogue values are unavailable, without filling them in from memory. The cell checks the final answer for parallaxes and spectral types that occur in no successful tool output.

````python
UNAVAILABLE_WORDING = (r"unavailable|not available|could not|couldn't|unable|no data|failed|time[sd]? ?out|"
                       r"did not respond|not (?:be )?retriev")
HEDGE_OR_ATTRIBUTION = re.compile(
    r"general knowledge|background knowledge|from memory|my (?:training|knowledge)|\brecall\b|literature|"
    r"commonly (?:cited|quoted|described|given|reported)|typically|well[- ]known|published|previously reported|"
    r"(?:could|can)(?:not|n't)(?: be)? (?:verif|confirm)|unverified|not (?:verified|confirmed)|"
    r"not (?:from|retrieved from|returned by) (?:SIMBAD|a tool|the tools?)|based on (?:my|general|prior|training)|"
    r"Hipparcos|Gaia|van Leeuwen", re.IGNORECASE)


def sentences(text: str) -> list[str]:
    """Sentences, bullet items and table rows of a text (decimals such as 6.55 are not split)."""
    return [part.strip() for part in re.split(r"(?<!\d)\.(?!\d)|[!?\n]", plain(text)) if part.strip()]


def untraceable_catalogue_values(run: dict) -> list[dict]:
    """Angular values in mas (parallaxes) and spectral types in the final answer that no successful tool output
    contains, with the sentence they occur in and any hedging or attribution wording in that sentence."""
    sources = physical_evidence(run)
    source_mas = [q["value"] for q in quantities(sources) if q["unit"] == "mas"]
    source_types = {canonical_spectral_type(s) for s in spectral_types_in(sources)}
    found = []
    for sentence in sentences(run["final_answer"] or ""):
        wording = sorted({match.group().lower() for match in HEDGE_OR_ATTRIBUTION.finditer(sentence)})
        values = [f"{q['text']} mas" for q in quantities(sentence)
                  if q["unit"] == "mas" and not number_supported(q["text"], source_mas)]
        values += [f"spectral type {s}" for s in spectral_types_in(sentence) if canonical_spectral_type(s) not in source_types]
        found += [{"value": value, "sentence": sentence, "wording": wording} for value in values]
    return found


def classify_f2(run: dict) -> tuple[str, str, list[str]]:
    """Verdict, short outcome and supporting detail lines for the SIMBAD-outage run."""
    answer = run["final_answer"] or ""
    simbad_attempts = [c for c in run["tool_calls"] if c["tool"] == "query_simbad"]
    arxiv_ok = bool(successful_results(run, "search_arxiv"))
    if not simbad_attempts:
        return "**Not triggered:** the model never called `query_simbad`.", "not triggered (no SIMBAD call)", []
    if any(not c["is_error"] for c in simbad_attempts):
        return "**Not triggered:** a SIMBAD call succeeded despite the 1 ms timeout.", "not triggered", []
    if run["status"] != "answered":
        return f"**No final answer** ({run['status']}); manual intervention required.", "manual intervention required", []
    untraceable = untraceable_catalogue_values(run)
    if untraceable:
        wording = sorted({w for item in untraceable for w in item["wording"]})
        verdict = (f"**The loop survived, but the answer quotes catalogue values that no tool returned.** SIMBAD never "
                   f"answered ({len(simbad_attempts)} attempt(s)), yet the final answer gives "
                   f"{', '.join(dict.fromkeys(item['value'] for item in untraceable))}. These can only come from the "
                   "model's memory, contrary to the system prompt. "
                   + (f"The sentences that contain them use the wording {', '.join(repr(w) for w in wording)}."
                      if wording else "No hedging or attribution wording (e.g. 'from memory', 'literature', "
                      "'Hipparcos') appears in those sentences."))
        details = ["* **Sentences with untraceable values:**"] + [
            f"    * \"{sentence[:220]}\"" for sentence in dict.fromkeys(item["sentence"] for item in untraceable)]
        return verdict, "answered, but quoted catalogue values from memory", details
    unavailable_said = sorted({m.group().lower() for m in re.finditer(UNAVAILABLE_WORDING, plain(answer), flags=re.IGNORECASE)})
    if unavailable_said:
        verdict = (f"**Graceful degradation.** The loop did not crash: after {len(simbad_attempts)} failed SIMBAD attempt(s) "
                   f"the model {'answered the arXiv part and ' if arxiv_ok else ''}said that the catalogue data were "
                   f"unavailable (wording: {', '.join(repr(w) for w in unavailable_said)}), and the answer contains no "
                   "parallax or spectral type that a tool did not return. Completing the answer still needs manual "
                   "intervention (re-running once SIMBAD is reachable).")
        return verdict, "graceful degradation, partial answer", []
    return ("**Incomplete answer without explanation.** SIMBAD never answered, and the final answer neither gives the "
            "catalogue values nor says that they are unavailable; manual intervention required."), \
        "incomplete answer, failure not reported", []


f2_run = await get_run("failure_F2_simbad_unreachable", betelgeuse_question, server_env={"SIMBAD_TIMEOUT_S": "0.001"})
print_run_trace(f2_run)

if f2_run:
    answer = f2_run["final_answer"] or ""
    verdict, failure_outcomes["F2"], details = classify_f2(f2_run)
    update_failure_doc("F2", "\n".join([
        observation_header(f2_run), "",
        "* **Tool calls:**", call_summary(f2_run),
        f"* **Recovery:** {verdict}", *details,
        "* **Final answer (excerpt):**", "", quote(answer)]))
````

### F3 — Agent turn budget exhausted

The same compound question with a budget of a single model call. The first reply should request tools, because the model has no data yet. The loop executes those calls and then stops before the model can read the results.

````python
f3_run = await get_run("failure_F3_turn_budget", betelgeuse_question, max_turns=1)
print_run_trace(f3_run)

if f3_run:
    if f3_run["final_answer"]:
        verdict = "**Not triggered:** the model answered within one call (without using the tools)."
        failure_outcomes["F3"] = "not triggered"
    else:
        fetched = {c["tool"] for c in f3_run["tool_calls"] if not c["is_error"]}
        verdict = (f"**No answer; manual intervention required.** The loop stopped with status *{f3_run['status']}* after "
                   f"executing {len(f3_run['tool_calls'])} tool call(s) whose results the model never saw. The user receives "
                   + ("nothing, although the data from both tools had already been fetched."
                      if {"query_simbad", "search_arxiv"} <= fetched else "nothing."))
        failure_outcomes["F3"] = "manual intervention required"
    update_failure_doc("F3", "\n".join([
        observation_header(f3_run), "",
        "* **Tool calls:**", call_summary(f3_run),
        f"* **Recovery:** {verdict}"]))
````

### F4 — Silent retrieval failure: "most recent" is not "most relevant"

No new run is needed: this failure is already present in the main runs of Sections 3 and 4. `search_arxiv` returns the newest abstracts that contain the keyword, as the question requests, but a paper can mention the object in a single sentence while being about something else. Nothing fails visibly, so the model may build an "evolutionary status" summary on papers that have nothing to say about it. Below, each returned paper is classified as *main subject* (the object's name or an alias such as α Ori in the title, or at least two mentions in the abstract) or *passing mention*. The result is then linked to the evolutionary-claim and claim-by-claim checks of Section 5.3.

````python
OBJECT_NAME_PATTERNS = {FIRST_OBJECT: r"betelgeuse|\b(?:alpha|α|alf)\.? ?ori(?:onis)?\b",
                        SECOND_OBJECT: r"proxima"}


def relevance_table(run: dict | None, name: str) -> pd.DataFrame:
    """Classify each returned paper by how prominently the object (or an alias) features in its title and abstract."""
    pattern = OBJECT_NAME_PATTERNS.get(name, re.escape(name))
    rows = []
    for paper in returned_papers(run):
        mentions = len(re.findall(pattern, plain(paper["abstract"]), flags=re.IGNORECASE))
        in_title = bool(re.search(pattern, plain(paper["title"]), flags=re.IGNORECASE))
        rows.append({"arXiv ID": paper["arxiv_id"], "title": paper["title"], "category": paper["primary_category"],
                     "mentions in abstract": mentions, "object in title": in_title,
                     "role of the object": "main subject" if in_title or mentions >= 2 else "passing mention"})
    return pd.DataFrame(rows)


f4_lines = []
f4_passing_total = 0
for name, run in runs_by_object.items():
    table = relevance_table(run, name)
    display(Markdown(f"**{name}**"))
    display(table)
    if run is None:
        f4_lines.append(f"* **{name}:** no run available.")
        continue
    if table.empty:
        f4_lines.append(f"* **{name}:** `search_arxiv` returned no papers in this run.")
        continue
    passing = table[table["role of the object"] == "passing mention"]
    f4_passing_total += len(passing)
    claims = claim_tables.get(name)
    verifier = ("n/a (no verifier result)" if claims is None else
                f"{int((claims['verdict'] != 'SUPPORTED').sum())} of {len(claims)} claims not supported")
    terms = evolution_terms[(evolution_terms["object"] == name) & ~evolution_terms["found in abstracts/SIMBAD"]]
    f4_lines.append(f"* **{name}:** {len(passing)} of {len(table)} returned papers mention the object only in passing"
                    + (": " + "; ".join(f"arXiv:{r['arXiv ID']} *{r['title']}*" for _, r in passing.iterrows()) if len(passing) else "")
                    + ". Evolution-related statements in the summary with no counterpart in the sources: "
                    + (", ".join(terms["statement in answer"]) if len(terms) else "none")
                    + f"; claim-by-claim check: {verifier}.")

f4_papers_total = sum(len(returned_papers(run)) for run in runs_by_object.values())
failure_outcomes["F4"] = (f"triggered silently ({f4_passing_total} passing-mention paper{'s' if f4_passing_total != 1 else ''})"
                          if f4_passing_total
                          else "not triggered in this run" if f4_papers_total else "not observed (no papers retrieved)")
if betelgeuse_run or second_run:
    reference_run = betelgeuse_run or second_run
    header = (f"*Observed in the main runs of `week05_ly.ipynb` (Sections 3-4, recorded {reference_run['started_utc']}, "
              f"model `{reference_run['model']}`).*")
    if f4_passing_total:
        verdict = ("**No error is raised, so the agentic loop cannot detect or recover from this failure by itself.** "
                   "Only the post-hoc comparison of the summary with the abstracts (Section 5.3) shows whether the model "
                   "drew conclusions from papers that are not about the object. Correcting it requires manual "
                   "intervention (a different query or a relevance filter).")
    elif f4_papers_total:
        verdict = "**Not triggered in this run:** every returned paper has the object as its main subject."
    else:
        verdict = "**Not observed:** no papers were returned, so the retrieval failure could not be assessed."
    update_failure_doc("F4", "\n".join([header, "", *f4_lines, f"* **Recovery:** {verdict}"]))
````

## 7. Run checks and summary

````python
def all_passed(table: pd.DataFrame) -> bool:
    """True if a check table is non-empty and every row passed."""
    return bool(len(table)) and bool(table["passed"].all())


citations_ok = all(runs_by_object.values()) and not any(
    grounding[name]["citations"]["cited but never returned"] for name in runs_by_object)
no_contradictions = all(table is not None and not (table["verdict"] == "CONTRADICTED").any()
                        for table in claim_tables.values())

run_checks = pd.DataFrame([
    ("MCP server starts as a subprocess and completes the handshake", bool(mcp_tools)),
    ("both tools advertised with complete JSON Schemas", schemas_complete),
    ("valid direct calls succeed; invalid arguments rejected with isError", valid_calls_succeeded and invalid_calls_rejected),
    (f"{FIRST_OBJECT}: both tools chained and an answer produced", all_passed(betelgeuse_chain)),
    (f"{SECOND_OBJECT}: both tools chained and an answer produced", all_passed(second_chain)),
    ("SIMBAD tool output agrees with the SIMBAD web interface", simbad_crosscheck_passed),
    ("physical sanity checks ran and none failed", sanity_failures == 0 and bool((sanity_checks["verdict"] == "PASS").any())),
    ("every arXiv ID cited in the answers was returned by search_arxiv", citations_ok),
    ("evolutionary class in both answers matches the SIMBAD luminosity class", stage_consistent),
    ("claim-by-claim verification available, with no CONTRADICTED claim", no_contradictions),
    ("failure modes F1-F4 written to week05_failures_ly.md",
     not re.search(r"Not yet observed", FAILURES_MD.read_text(encoding="utf-8"))),
], columns=["check", "passed"])
run_checks
````

````python
def physics_summary() -> str:
    """Summary of the physical sanity checks, worded according to their verdicts, object by object."""
    def verdicts(name: str, prefix: str) -> list[str]:
        rows = sanity_checks[(sanity_checks["object"] == name) & sanity_checks["check"].str.startswith(prefix)]
        return list(rows["verdict"])

    def usable(record: dict) -> bool:
        return bool(record.get("parallax_mas")) and record["parallax_mas"] > 0 and record.get("parallax_error_mas") is not None

    parts = []
    betelgeuse, proxima = record_for(FIRST_OBJECT), record_for(SECOND_OBJECT)
    if usable(betelgeuse):
        d_b, sd_b = parallax_distance(betelgeuse["parallax_mas"], betelgeuse["parallax_error_mas"])
        radii = [(theta / 2000) * d_b * AU_IN_SOLAR_RADII for theta in BETELGEUSE_THETA_MAS.values()]
        parts.append(f"Betelgeuse's parallax gives {d_b:.0f} ± {sd_b:.0f} pc (fractional error {100 * sd_b / d_b:.1f} %"
                     + (", large enough for 1/parallax to be biased" if sd_b / d_b > 0.10 else "") + "), which "
                     + ("agrees within 2σ with both independent estimates (168 and 222 pc)"
                        if all(v == "PASS" for v in verdicts(FIRST_OBJECT, "distance vs."))
                        else "disagrees by more than 2σ with at least one independent estimate (168 or 222 pc)")
                     + f"; the implied radius is {min(radii):.0f}-{max(radii):.0f} R_sun, "
                     + ("as expected for a red supergiant."
                        if all(v == "PASS" for v in verdicts(FIRST_OBJECT, "radius from"))
                        else "outside the expected red-supergiant range."))
    else:
        parts.append("Betelgeuse's checks could not run because its SIMBAD parallax is missing from the run.")
    if usable(proxima):
        d_p, _ = parallax_distance(proxima["parallax_mas"], proxima["parallax_error_mas"])
        parts.append(f"Proxima's parallax gives {d_p:.4f} pc = {d_p * PC_IN_LIGHT_YEARS:.3f} ly, "
                     + ("closer than α Cen AB, as the nearest star must be."
                        if verdicts(SECOND_OBJECT, "nearest-star") == ["PASS"] else "which fails the nearest-star test."))
    else:
        parts.append("Proxima's checks could not run because its SIMBAD parallax is missing from the run.")
    return " ".join(parts)


def summary_markdown() -> str:
    """Summary of the whole notebook, generated from the computed results."""
    def flagged(name: str) -> str:
        table = claim_tables.get(name)
        return "n/a" if table is None else f"{int((table['verdict'] != 'SUPPORTED').sum())} of {len(table)}"

    def unsupported_terms(name: str) -> int:
        return int(((evolution_terms["object"] == name) & ~evolution_terms["found in abstracts/SIMBAD"]).sum())

    tokens = sum(run["input_tokens"] + run["output_tokens"] for run in (betelgeuse_run, second_run) if run)
    retrieval = ("The recency-ordered keyword search returned papers that mention the object only in passing (F4), so "
                 "an evolutionary-status summary has little support in the retrieved abstracts and is easily filled in "
                 "from the model's background knowledge." if f4_passing_total else
                 "Every retrieved paper had the object as its main subject in this run." if f4_papers_total else
                 "No papers were retrieved, so the summaries could not be compared with abstracts.")
    return "\n".join([
        "### Summary",
        "",
        f"* **Pipeline.** The MCP server advertises `query_simbad` and `search_arxiv` with complete input and output "
        f"schemas (check passed: {schemas_complete}) and rejects invalid arguments with `isError` results "
        f"({invalid_calls_rejected}). With `{LLM_MODEL}` the agent chained both tools for {FIRST_OBJECT} "
        f"({all_passed(betelgeuse_chain)}) and {SECOND_OBJECT} ({all_passed(second_chain)}), "
        f"using {tokens} tokens in total for the two main runs.",
        f"* **SIMBAD cross-check.** {int(simbad_crosscheck['agree'].sum())} of {len(simbad_crosscheck)} fields agree "
        "with the SIMBAD web interface.",
        f"* **Physical sanity checks.** {len(sanity_checks)} checks, {sanity_failures} failed. {physics_summary()}",
        f"* **Summary vs. abstracts.** Evolution-related statements with no counterpart in the sources: "
        f"{FIRST_OBJECT} {unsupported_terms(FIRST_OBJECT)}, {SECOND_OBJECT} {unsupported_terms(SECOND_OBJECT)}. "
        "Evolutionary class consistent with the SIMBAD luminosity class: "
        + ", ".join(f"{row['object']} {'n/a' if row['stated in final answer'] is None else row['stated in final answer']}"
                    for _, row in stage_check.iterrows()) + ". "
        f"Claims the verifier could not support: {FIRST_OBJECT} {flagged(FIRST_OBJECT)}, "
        f"{SECOND_OBJECT} {flagged(SECOND_OBJECT)}. {retrieval}",
        "* **Failure modes.** " + "; ".join(f"{key}: {value}" for key, value in failure_outcomes.items()) + ". "
        "Errors reported by the server reach the model as `is_error` tool results instead of crashing the loop, so "
        "recovery from them depends on the model; a loop that runs out of turns and a silent retrieval failure cannot "
        "be detected from inside the loop.",
    ])


display(Markdown(summary_markdown()))
````
